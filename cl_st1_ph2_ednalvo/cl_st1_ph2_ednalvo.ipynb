## Processing 1

### Rename text files using 5-digit essay IDs

````python
from pathlib import Path
import re

dataset_dir = Path("corpus/01_dataset")

for txt_file in dataset_dir.glob("*.txt"):
    match = re.match(r"^(\d{5})", txt_file.name)

    if not match:
        print(f"Skipped: {txt_file.name}")
        continue

    new_name = f"{match.group(1)}.txt"
    new_path = txt_file.with_name(new_name)

    if txt_file == new_path:
        print(f"Already renamed: {txt_file.name}")
        continue

    if new_path.exists():
        print(f"Skipped because target already exists: {txt_file.name} -> {new_name}")
        continue

    txt_file.rename(new_path)
    print(f"Renamed: {txt_file.name} -> {new_name}")
````
